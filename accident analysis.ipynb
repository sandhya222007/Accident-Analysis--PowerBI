```python
import pandas as pd
df = pd.read_csv("AccidentInsight_500_Rows.csv")
df.head()
```

```python
print("Shape:", df.shape)
```

```python
print("\nColumns:")
print(df.columns)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows:", df.duplicated().sum())
print("Shape:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["Weather"] = df["Weather"].fillna(df["Weather"].mode()[0])

df["Road_Condition"] = df["Road_Condition"].fillna(
    df["Road_Condition"].mode()[0]
)
df["Driver_Age"] = df["Driver_Age"].fillna(
    df["Driver_Age"].mean()
)
```

```python
print(df.isnull().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
print(df.shape)
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Accident_Severity", color="red")

plt.title("Accident Severity Distribution")
plt.xlabel("Accident Severity")
plt.ylabel("Number of Accidents")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Weather")

plt.title("Accidents by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Number of Accidents")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Road_Condition", color="lightgreen")

plt.title("Accidents by Road Condition")
plt.xlabel("Road Condition")
plt.ylabel("Number of Accidents")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Accident_Severity", color="salmon")

plt.title("Accident Severity Distribution")
plt.xlabel("Accident Severity")
plt.ylabel("Number of Accidents")

plt.show()
```

```python
df["Month"] = df["Date"].dt.month

monthly = df.groupby("Month").size()

plt.figure(figsize=(9,5))

plt.plot(monthly.index, monthly.values, marker="o", color="purple")

plt.title("Monthly Accident Trend")
plt.xlabel("Month")
plt.ylabel("Number of Accidents")

plt.xticks(range(1,13))

plt.show()
```

```python
plt.figure(figsize=(10,7))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
